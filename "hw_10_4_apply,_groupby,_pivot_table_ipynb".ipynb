## 1 гіпотеза . В які дні тижня бльше всього купують?

Перетворимо Date у формат дати та створимо день тижня

```python
df['Date'] = pd.to_datetime(df['Date'])

df['Day_of_week'] = df['Date'].dt.day_name()
```

Прахуємо кількість покупок та середній чек

```python
purchases_by_day = df.groupby('Day_of_week').agg(
    invoices = ('Invoice ID', 'count'),
    average_total=('Total', 'mean')
).sort_values('invoices', ascending=False)

purchases_by_day
```

```python
top_day = purchases_by_day['invoices'].idxmax()
max_invoices = purchases_by_day['invoices'].max()

print(f'Найбільше покупок здійснюють у день: {top_day}')
print(f'Кількість покупок: {max_invoices}')
```

Висновок: найбільше покупок здійснюють у суботу: 164 інвойси. Субота також має найвищий середній чек, який становить приблизно 342,20. Найменше покупок здійснюють у понеділок: 125 інвойсів, а середній чек становить приблизно 303,19. Отже, припущення про підвищену купівельну активність у вихідні підтвердилося лише частково, оскільки субота є найактивнішим днем, але в неділю кількість покупок нижча, ніж у деякі будні.
##Бізнес-рекомендація:
 E суботу магазину доцільно збільшувати кількість працівників, контролювати товарні запаси та проводити акції для збільшення продажів. Для понеділка можна розробити спеціальні знижки або промопропозиції, щоб стимулювати попит у день із найнижчою кількістю покупок.


##Гіпотеза 2:
покупці типу Member здійснюють більше покупок, купують більше товарів та мають вищий середній чек, ніж покупці типу Normal.

Вони також купують у середньому трохи більше товарів: 5,56 проти 5,46. Загальна виручка від покупців Member становить 164 223,44, що на 5 480,14 більше, ніж від покупців Normal. Отже, учасники програми лояльності є дещо ціннішими для магазину, однак програма поки що не створює великої різниці в купівельній поведінці.
## Бізнес рекомендація
магазину доцільно продовжувати розвивати програму лояльності, оскільки покупці Member мають вищий середній чек і приносять більшу виручку. Водночас варто посилити переваги програми, наприклад запровадити персональні знижки, накопичувальні бонуси або спеціальні пропозиції, щоб збільшити різницю між поведінкою Member і Normal.

##Гіпотеза 3:
 деякі категорії товарів мають велику кількість покупок, але нижчий середній рейтинг. Це може свідчити про проблеми з якістю товарів, асортиментом або обслуговуванням.

Створюєм таблицю для аналізу

```python
category_analysis = df.groupby('Product line').agg(
    invoices=('Invoice ID', 'count'),
    average_rating=('Rating', 'mean'),
    average_total=('Total', 'mean')
).round(2)

category_analysis
```

Знайдемо категорію з найнижчим рейтингом

```python
lowest_rating_category = category_analysis['average_rating'].idxmin()
lowest_rating = category_analysis['average_rating'].min()

print(f'Категорія з найнижчим середнім рейтингом: {lowest_rating_category}')
print(f'Середній рейтинг: {lowest_rating:.2f}')
```

Перевіримо  чи високий попит у неї

```python
category_invoices = category_analysis.loc[
    lowest_rating_category,
    'invoices'
]

print(f'Кількість інвойсів: {category_invoices}')
```

Поріняємо тепер із середньою кількістю інвойсів

```python
average_invoices = category_analysis['invoices'].mean()

print(f'Середня кількість інвойсів серед категорій: {average_invoices:.2f}')
```

Виводимо перевірку програмно

```python
if category_invoices > average_invoices:
    print(
        'Гіпотеза підтвердилася: категорія з найнижчим рейтингом '
        'має кількість інвойсів вище середнього.'
    )
else:
    print(
        'Гіпотеза не підтвердилася: категорія з найнижчим рейтингом '
        'не має кількості інвойсів вище середнього.'
    )
```

##Висновок:
гіпотеза не підтвердилася. Категорія з найнижчим середнім рейтингом має 160 інвойсів, тоді як середня кількість інвойсів серед усіх категорій становить 166,67. Отже, кількість покупок у цій категорії приблизно на 4% нижча за середню. Це означає, що категорія з найнижчим рейтингом не має високого попиту відносно інших категорій.
##Бізнес рекомендація
магазину варто додатково дослідити причини низького рейтингу цієї категорії, зокрема якість товарів, ціни та обслуговування. Оскільки попит також нижчий за середній, покращення пропозиції може одночасно підвищити задоволеність покупців і кількість продажів.
