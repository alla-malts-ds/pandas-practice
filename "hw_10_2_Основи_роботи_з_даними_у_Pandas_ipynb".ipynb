## Гіпотеза 2
У Mandalay можуть частіше купувати товари з дорожчих категорій. Для перевірки потрібно порівняти кількість продажів і середній gross income за Product line у Mandalay та Yangon. Якщо гіпотеза підтвердиться, у Yangon можна активніше просувати прибуткові товарні категорії.

```python
product_analysis = (
    cities
    .groupby(['City', 'Product line'])
    .agg(
        invoices=('Invoice ID', 'count'),
        avg_unit_price=('Unit price', 'mean'),
        avg_total=('Total', 'mean'),
        avg_gross_income=('gross income', 'mean')
    )
)

product_analysis['share_percent'] = (
    product_analysis['invoices']
    / product_analysis.groupby(level='City')['invoices'].transform('sum')
    * 100
)

product_analysis.round(2)
```

## Висновок за гіпотезою 2
Гіпотеза підтвердилася частково. Найбільшу перевагу Mandalay забезпечує категорія Health and beauty: середній gross income становить 17.95 проти 12.76 у Yangon, а її частка продажів також вища: 15.96% проти 13.82%.

## Гіпотеза 3
У Mandalay більша частка покупок може припадати на клієнтів Member, а їхній середній чек і gross income можуть бути вищими, ніж у клієнтів Yangon. Це може частково пояснювати вищий середній gross income у Mandalay.

Для перевірки гіпотези потрібно порівняти частку інвойсів Member, їхній середній Total і середній gross income у двох містах. Якщо гіпотеза підтвердиться, у Yangon можна активніше розвивати програму лояльності.

```python
# 1. Залишаємо два потрібні міста
cities = df[df['City'].isin(['Mandalay', 'Yangon'])].copy()

# 2. Рахуємо показники за містом і типом клієнта
customer_analysis = (
    cities
    .groupby(['City', 'Customer type'])
    .agg(
        invoices=('Invoice ID', 'count'),
        avg_total=('Total', 'mean'),
        avg_gross_income=('gross income', 'mean')
    )
    .reset_index()
)

# 3. Розраховуємо частку інвойсів кожного типу клієнта в місті
customer_analysis['share_percent'] = (
    customer_analysis['invoices']
    / customer_analysis.groupby('City')['invoices'].transform('sum')
    * 100
)

# 4. Округлюємо результати
customer_analysis = customer_analysis.round(2)

customer_analysis
```

Гіпотеза не підтвердилася повністю. У Mandalay частка інвойсів Member, їхній середній чек і gross income лише трохи вищі, ніж у Yangon. Різниця є незначною, тому тип клієнта навряд чи є основною причиною вищого gross income у Mandalay.
